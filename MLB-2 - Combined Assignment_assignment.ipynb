**How to do it**

For indexing, remember counting starts at zero and negative numbers count from the end. For slicing, the pattern is start, stop and step, and the stop position is not included.
```python
# Create one 1D, one 2D and one 3D array, and write at least 5 indexing or slicing examples for each
# 1D array
a1 = np.array([5,10,15,20,25,30,35,40,45,50])
print("1D Indexing & Slicing:")
print("Single element:",a1[2])
print("Range:", a1[1:4])
print("Step:", a1[::2])
print("Reverse order:", a1[::-1])
print("Condition:", a1[a1 > 30])
# 2D array
a2 = np.array([[2,4,6],
              [8,10,12],
              [14,16,18]])
print("2D Indexing & Slicing:")
print("Single element:", a2[1,1])
print("Whole row:", a2[0,:])
print("Whole column:", a2[:,1])
print("Range:", a2[0:2, 1:3])
print("Reverse rows:", a2[::-2, :])
# 3D array
a3 = np.array([[[1,3],[2,4],
               [5,7],[6,8]]])
print("3D Indexing & Slicing:")
print("Single element:", a3[0,1,0])
print("Whole first 2D matrix:", a3[0, :, :])
print("Specific column:", a3[:, :, 0])
print("Range:", a3[0:2,:, :])
print("Reverse:", a3[::-1, :, :])
```

## Question 12: Find and mitigate outliers

**Tasks**

- Find outliers in the age column of the diabetes dataset, then mitigate them using both trimming and capping.
- Find outliers in the price column of the house dataset, then mitigate them using both trimming and capping.

**How to do it**

Detect first: draw a box plot of the column and look at its summary statistics. Then use the IQR rule: find the 25th percentile (Q1) and the 75th percentile (Q3), calculate IQR as Q3 minus Q1, set the lower limit as Q1 minus 1.5 times IQR and the upper limit as Q3 plus 1.5 times IQR. Any value outside the limits is an outlier; count them. Trimming means removing the rows that are outside the limits, so the dataset becomes smaller. Capping means keeping all rows but replacing values below the lower limit with the lower limit and values above the upper limit with the upper limit. Work on copies of the data so the original stays untouched. Compare the number of rows and the box plots before and after each method. For the house price column, also check whether there are unrealistic values such as zero, and write one line on how you decided to handle them. If a column has no outliers, state that clearly and show the evidence.

### Outliers in the age column of the diabetes dataset,then mitigate them using both trimming and capping

```python
# Draw Boxplot first and check summary statistics
sns.boxplot(x=diabetes['age'])
plt.title("Original Diabetes 'Age' Boxplot")
plt.show()

print("Summary Statistics:")
print(diabetes['age'].describe())
# Calculating limit using IQR rules
Q1_age = diabetes['age'].quantile(0.25)
Q3_age = diabetes['age'].quantile(0.75)

print("25th percentile, Q1: ",Q1_age)
print("75th percentile, Q3: ",Q3_age)

IQR = Q3_age - Q1_age
print("IQR: ",IQR)
upper_limit = Q3_age + 1.5 * IQR
lower_limit = Q1_age - 1.5 * IQR
outliers_limit = diabetes[(diabetes['age'] < lower_limit) | (diabetes['age'] > upper_limit)]
print("Upper limit: ",upper_limit)
print("Lower limit: ",lower_limit)
print(f"Total Outliers Count: {len(outliers_limit)}")
```

#### **Evidence: Lower limit = -30.0 and Upper limit = 114.0. All values in the age column fall between 0.08 and 80.0. Therefore, there are 0 outliers in the diabetes dataset's age column

```python
# Mitigation (Triming & Capping)
# Trimming
print("Before trimming:",diabetes.shape)

new_diabetes = diabetes[(diabetes['age'] <= upper_limit) & (diabetes['age'] >= lower_limit)]
print("After trimming:",new_diabetes.shape)
sns.boxplot(new_diabetes['age'])
plt.title("Diabetes 'Age' Boxplot(After Trimming)")
plt.show()
# Capping
new_df_cap = diabetes.copy()

new_df_cap['age'] = np.where(new_df_cap['age'] > upper_limit, upper_limit,
                    np.where(new_df_cap['age'] < lower_limit, lower_limit,
                             new_df_cap['age']))

print("Before capping:",diabetes.shape)
print("After capping:",new_df_cap.shape)

sns.boxplot(new_df_cap['age'])
plt.title("Diabetes 'Age' Boxplot(After Capping)")
plt.show()
```

### Outliers in the price column of the house dataset, then mitigate them using both trimming and capping
