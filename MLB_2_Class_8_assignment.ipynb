```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
titanic = pd.read_csv('/kaggle/input/datasets/lovishbansal123/titanic-dataset/Titanic-Dataset.csv')
iris = pd.read_csv('/kaggle/input/datasets/himanshunakrani/iris-dataset/iris.csv')
tips = pd.read_csv('/kaggle/input/datasets/saurabhbadole/restaurant-tips-dataset/tips.csv')
```

# Univariate Analysis

# 1. Pie chart

```python
tips
```

```python
data = tips['smoker'].value_counts()

plt.figure(figsize=(5, 5))
plt.pie(data, labels=data.index, autopct='%1.2f%%', startangle=90)
plt.title('Smoker vs Non-Smoker')
plt.show()
```

# 2. Count plot

```python
sns.countplot(data = tips, x = 'sex')
plt.title('Count plot on Gender Distribution')
plt.xlabel('Sex')
plt.ylabel('Count')
plt.show()
```

# 3. Kde + Rug plot

```python
sns.kdeplot(data = tips, x = 'total_bill')
plt.title('Density distribution of toal bill')
plt.show()
```

```python
sns.kdeplot(tips['total_bill'])
sns.rugplot(tips['total_bill'])
plt.title('KDE and rug pot on total bill')
plt.show()
```

# 4. Violin plot

```python
sns.violinplot(data = tips['tip'])
plt.title('Violin plot on tips')
plt.show()
```

# 5. Box plot

```python
sns.boxplot(data = tips, x = 'size')
plt.title('Box plot of Size')
plt.show()
```

# 6. Boxen plot

```python
sns.boxenplot(data = tips,x = 'tip')
plt.title('Boxen plot on tip column')
plt.show()
```

# 7. Histogram

```python
sns.histplot(data = tips, x = 'total_bill')
plt.title('Distribution of Total Bill')
plt.show()
```

# Bivariate Analysis

# 1. Scatter plot

```python
sns.scatterplot(data = titanic, x = 'Age', y = 'Survived')
plt.title('Scatter plot on age and survived')
plt.show()
```

# 2. Line plot

```python
sns.lineplot(data = titanic, x = 'Age', y = 'Fare')
plt.title('Line plot on age and fare')
plt.show()
```

# 3. Joint plot

```python
sns.jointplot(data = titanic, x = 'Pclass', y = 'Fare')
plt.title('Joint plot between Pclass and Fare')
plt.show()
```

# 4. Hexbin plot

```python
plt.hexbin(titanic['Age'], titanic['Fare'], gridsize=25, cmap='Reds')

plt.colorbar(label='count in bin')
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Hexbin Plot: Age vs Fare")

plt.show()
```

# 5. 2D kde plot

```python
sns.kdeplot(data = titanic, x='SibSp', y='Parch', fill = True)
plt.xlabel("SibSp")
plt.ylabel("Parch")
plt.title("2D KDE Plot between SibSp and Parch")
plt.show()
```

# 6. Box plot

```python
sns.boxplot(data = titanic, x = 'Survived', y = 'Pclass', palette = 'Set2')
plt.title('Boxplot on survived according to pclass')
plt.show()
```

# 7. Violin plot

```python
sns.violinplot(data = titanic, x = 'Survived', y = 'Age', palette = 'Set2')
plt.title('Violin plot between Survived and Age of Titanic dataset')
plt.show()
```

# 8. Swarm plot

```python
sns.swarmplot(data = titanic, x='Sex',y='Fare',hue='Sex')
plt.title('Swarm plot between Sex and Fare of Titanic dataset')
plt.show()
```

# 9. Count plot

```python
sns.countplot(data = titanic, x = 'Embarked', hue = 'Survived')
plt.title('Count plot on embarked and survived passenger')
plt.show()
```

# Multivariate Analysis

# 1. Scatter plot

```python
sns.scatterplot(data = iris, x = 'sepal_length', y = 'sepal_width', hue = 'species')
plt.title('Scatter plot on sepal length and sepal width to find species')
plt.show()
```

# 2. Pair plot

```python
sns.pairplot(iris, hue = 'sepal_length')
plt.show()
```

# 3. Heatmap

```python
sns.heatmap(iris.corr(numeric_only = True))
plt.title('Heatmap showing correlation between iris features')
plt.show()
```

# 4. FacetGrid plot

```python
g = sns.FacetGrid(titanic,col = 'Pclass',row = 'Embarked')
g.map_dataframe(sns.histplot,x = 'Fare')
plt.show()
```

# 5. PairGrid

```python
g = sns.PairGrid(titanic,hue='Survived')
g.map_diag(sns.histplot)
g.map_upper(sns.scatterplot)
g.map_lower(sns.kdeplot)
g.add_legend()
plt.show()
```

# 6. Joint plot

```python
sns.jointplot(data = titanic, x = 'Age', y = 'SibSp', hue = 'Sex')
plt.show()
```

# 7. Boxen plot

```python
sns.boxenplot(data = titanic, x = 'Pclass', y = 'Age', hue = 'Survived')
plt.show()
```
